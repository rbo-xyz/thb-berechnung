```python
from utils.imports import import_csv
import pandas as pd
from pathlib import Path
```

```python
base_path = Path(r"C:\Users\r.boppart\FHNW\E-BSc Feldkurs 2025 - VP 5220 GeoSensorik&Monitoring_M365 - Madrisa\500_Auswertung\10_THB-Messungen\Auswertung_Python")

csv_first = []
csv_second = []
parent_folders = []
folder_paths = []
csv_first_names = []
csv_second_names = []

for folder in sorted([f for f in base_path.iterdir() if f.is_dir()]):
       
       ## Ignoriere den Ordner "_all-data"
       if folder.name == "_all-data":
            continue
       csv_files = sorted([f for f in folder.glob("*.csv")])
       if len(csv_files) >= 2:
            csv_first.append(csv_files[0])
            csv_second.append(csv_files[1])
            csv_first_names.append(csv_files[0].name)
            csv_second_names.append(csv_files[1].name)
            parent_folders.append(folder.name)
            folder_paths.append(str(folder))
       else:
            print(f"Warnung: Weniger als 2 CSV-Dateien in {folder.name}")

for i in range(len(parent_folders)):

    final_a = csv_first[i].parent / "Messungen_ppm-Korr" / (csv_first[i].stem + "_korr" + csv_first[i].suffix)
    final_b = csv_second[i].parent / "Messungen_ppm-Korr" / (csv_second[i].stem + "_korr" + csv_second[i].suffix)

    koor_folder = csv_first[i].parent / "Messungen_ppm-Korr"
    koor_folder.mkdir(parents=True, exist_ok=True)

    csv_first_new = str(csv_first[i])
    csv_second_new = str(csv_second[i])

    df_a = import_csv(csv_first_new)
    df_b = import_csv(csv_second_new)

    df_a.loc[df_a["Lage"] == "2", "V-Winkel"] = 400 - df_a["V-Winkel"]
    df_b.loc[df_b["Lage"] == "2", "V-Winkel"] = 400 - df_b["V-Winkel"]

    drop_col = ["Datum", "Uhrzeit", "Session", "Lage", "MessNr", "Hz-Winkel"]

    df_a = df_a.drop(columns=drop_col, errors="ignore")
    df_b = df_b.drop(columns=drop_col, errors="ignore")

    df_a.to_csv(final_a, mode="w", index=False)
    df_b.to_csv(final_b, mode="w", index=False)
```
